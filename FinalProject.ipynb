```python
df_clean = df_raw.drop_duplicates().copy()
df_clean.duplicated().sum() # to make sure that the duplicated is removed
```

```python
df_clean = df_clean.rename(columns={"Leght_m": "Length_m","Dai_cm": "Diameter_cm"})
df_raw = df_raw.rename(columns={"Leght_m": "Length_m","Dai_cm": "Diameter_cm"})
num_cols = ["Length_m", "Diameter_cm", "Ww_kg", "Dw_kg", "%Moisture","Dw","LogVolume_m3", "DwToWwRatio", "Altitude_m", "Slope_deg"]
```

```python
df_clean = df_clean.replace(["", " ", "NA", "N/A", "na", "null", "-"], np.nan)
df_clean[num_cols].isna().sum()
```

```python
df_clean.loc[df_clean["Length_m"] <= 0, "Length_m"] = np.nan
df_clean.loc[df_clean["Diameter_cm"] <= 0, "Diameter_cm"] = np.nan
df_clean.loc[df_clean["Ww_kg"] <= 0, "Ww_kg"] = np.nan
df_clean.loc[df_clean["Dw_kg"] <= 0, "Dw_kg"] = np.nan
df_clean.loc[df_clean["Dw"] <= 0, "Dw"] = np.nan
df_clean.loc[df_clean["LogVolume_m3"] <= 0, "LogVolume_m3"] = np.nan
df_clean.loc[(df_clean["%Moisture"] < 0),"%Moisture"] = np.nan
df_clean.loc[(df_clean["DwToWwRatio"] <= 0) | (df_clean["DwToWwRatio"] > 1),"DwToWwRatio"] = np.nan
df_clean.loc[df_clean["Altitude_m"] < 0, "Altitude_m"] = np.nan
df_clean.loc[(df_clean["Slope_deg"] < 0) | (df_clean["Slope_deg"] > 90),"Slope_deg"] = np.nan
```

```python
df_clean["Ww_temp"] = pd.qcut(df_clean["Ww_kg"],
                             q=4,
                             duplicates="drop")
df_clean["Diameter_cm"] = df_clean.groupby("Ww_temp", observed=False)["Diameter_cm"].transform(lambda x:x.fillna(x.median()))
df_clean.drop(columns=["Ww_temp"], inplace=True)
df_clean["Diameter_cm"].isna().sum()
```

```python
plt.figure()
plt.hist(df_clean['Diameter_cm'])
plt.title('Diameter Distribution')
plt.xlabel('Diameter_cm')
plt.ylabel('Frequency')
plt.show()
```

```python
df_clean["Diameter_temp"] = pd.qcut(df_clean["Diameter_cm"],
                              q=4,
                              duplicates="drop")
df_clean["LogVolume_m3"] = df_clean.groupby("Diameter_temp", observed=False)["LogVolume_m3"].transform(lambda x: x.fillna(x.median()))
df_clean.drop(columns=["Diameter_temp"], inplace=True)
df_clean["LogVolume_m3"].isna().sum()
```

```python
df_clean["Length_m"].isna().sum()
```

```python
df_clean["Vol_temp"] = pd.qcut(df_clean["LogVolume_m3"],
                               q=4,
                               duplicates="drop")
df_clean["Length_m"] = (df_clean.groupby("Vol_temp", observed=False)["Length_m"].transform(lambda x: x.fillna(x.median())))
df_clean.drop(columns=["Vol_temp"], inplace=True)
df_clean["Length_m"].isna().sum()
```
